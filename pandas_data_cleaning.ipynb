## Overview
This notebook cleans raw healthcare data for PostgreSQL loading and Power BI visualization.

## Cleaning Steps
1. Fix name capitalization
2. Remove duplicate 'Disharge Date' column
3. Convert date columns to datetime
4. Calculate Length of Stay
5. Remove duplicates
6. Export cleaned CSV

## Output
- `Healthcare_cleaned.csv` (54,966 rows, 17 columns)

```python
# Import Libraries
import pandas as pd
```

```python
# Load Data
df = pd.read_csv('/Users/grigorijgorevic/Desktop/Healthcare_dataset.csv')
```

```python
# Initial Data Exploration
print("= First 5 rows =")
print(df.head())

print("\n= Data Types =")
print(df.dtypes)

print("\n= Missing Values =")
print(df.isnull().sum())

print("\n= Basic Statistics =")
print(df.describe())

print(df.head(1))
```

```python
# Data Cleaning - Text Columns
# Fix name capitalization
df['Name'] = df['Name'].str.title()
print("Names fixed:", df['Name'].head(3))

# Remove duplicate typo column 'Disharge Date'
if 'Disharge Date' in df.columns:
    df = df.drop(columns=['Disharge Date'])
    print("Removed duplicate column 'Disharge Date'")

# Check gender values for any "Typo"
print(f"Unique genders: {df['Gender'].unique()}")
```

```python
# Data Cleaning - Date Columns
# Convert to datetime
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

# Create Length of Stay column
df['Length of Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
print(f"Length of Stay - Min: {df['Length of Stay'].min()}, Max: {df['Length of Stay'].max()}")
```

```python
# Remove Duplicates
print(f"Duplicates before: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Duplicates after: {df.duplicated().sum()}")
```

```python
# Final Summary & Validation
print("FINAL DATA SUMMARY")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Date range: {df['Date of Admission'].min()} to {df['Date of Admission'].max()}")
print(f"Avg Length of Stay: {df['Length of Stay'].mean():.1f} days")
print(f"Avg Billing Amount: ${df['Billing Amount'].mean():.2f}")
```

```python
# Export Cleaned Data
df.to_csv('Healthcare_cleaned.csv', index=False)
```
